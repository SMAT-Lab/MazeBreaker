```python
import pandas as pd
import os

# 获取文件列表
files = os.listdir('results')

results = []

for file in files:
    df = pd.read_csv('results/' + file)
    
    if df.shape[0] < 1: 
        # print(f"File {file} has no rows, skipping.")
        continue

    data_columns = df.columns[1:]  
    
    df['count_of_1'] = (df[data_columns] == 1).sum(axis=1)
    
    df_sorted = df.sort_values(by='count_of_1', ascending=False)

    if df_sorted.shape[0] >= 1:
        top1_row = df_sorted.iloc[0, 1:-1] 
        top1_asr = (top1_row == 1).sum() 
    else:
        top1_asr = 0  

    if df_sorted.shape[0] >= 5:
        top5_rows = df_sorted.iloc[:5, 1:-1] 
        top5_asr = (top5_rows == 1).any().sum()
    else:
        top5_rows = df_sorted.iloc[:, 1:-1]
        top5_asr = (top5_rows == 1).any().sum()

    results.append({
        'file': file,
        'Top1-ASR': top1_asr,
        'Top5-ASR': top5_asr,
        'sorted_data': df_sorted[['count_of_1']]
    })

for result in results:
    print(f"File: {result['file']}")
    print(f"Top1-ASR: {result['Top1-ASR']}")
    print(f"Top5-ASR: {result['Top5-ASR']}")
    # print(result['sorted_data'])
```
