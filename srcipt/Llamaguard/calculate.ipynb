```python
import pandas as pd
import os

# 获取文件列表
files = os.listdir('results')

# 存储每个文件的结果
results = []

# 遍历每个文件
for file in files:
    # 读取文件
    df = pd.read_csv('results/' + file)
    
    # 确保文件中有数据行
    if df.shape[0] < 1:  # 没有数据行时跳过该文件
        # print(f"File {file} has no rows, skipping.")
        continue
    
    # 跳过第一列数据
    data_columns = df.columns[1:]  # 跳过第0列
    
    # 计算每一行中1的个数
    df['count_of_1'] = (df[data_columns] == 1).sum(axis=1)
    
    # 1. 按照count_of_1列降序排序
    df_sorted = df.sort_values(by='count_of_1', ascending=False)

    # 确保至少有一行可以用于Top1的计算
    if df_sorted.shape[0] >= 1:
        # 选取1最多的一行，计算Top1-ASR
        top1_row = df_sorted.iloc[0, 1:-1]  # 选取最上面的一行 (跳过第0列和最后的count_of_1列)
        top1_asr = (top1_row == 1).sum()    # 计算Top1行有多少列是1
    else:
        top1_asr = 0  # 如果没有足够的行，Top1-ASR设置为0

    # 确保至少有五行可以用于Top5的计算
    if df_sorted.shape[0] >= 5:
        # 选取1最多的前五行，计算Top5-ASR
        top5_rows = df_sorted.iloc[:5, 1:-1]  # 选取前5行数据 (跳过第0列和最后的count_of_1列)
        # 对于每个问题，检查前5行中是否有一个是1，若有则计为成功
        top5_asr = (top5_rows == 1).any().sum()
    else:
        # 如果不足5行，只选取现有的所有行进行Top5-ASR计算
        top5_rows = df_sorted.iloc[:, 1:-1]
        top5_asr = (top5_rows == 1).any().sum()

    # 存储每个文件的结果
    results.append({
        'file': file,
        'Top1-ASR': top1_asr,
        'Top5-ASR': top5_asr,
        'sorted_data': df_sorted[['count_of_1']]
    })

# 最后统一输出结果
for result in results:
    print(f"File: {result['file']}")
    print(f"Top1-ASR: {result['Top1-ASR']}")
    print(f"Top5-ASR: {result['Top5-ASR']}")
    # print(result['sorted_data'])
```
